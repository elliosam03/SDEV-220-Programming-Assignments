```python
# 3.1 
seconds_per_minute = 60
minutes_per_hour = 60
seconds_per_hour = seconds_per_minute * minutes_per_hour
print ("Seconds per hour =",seconds_per_hour)

# 3.2
seconds_per_hour = 3600

# 3.3
hours_per_day = 24
seconds_per_day = seconds_per_hour * hours_per_day
print ("Seconds per day =",seconds_per_day)

# 3.4
seconds_per_day = 86400

# 3.5
print ("Floating-Point Division Answer =",seconds_per_day / seconds_per_hour)

# 3.6
print ("Integer Division Answer =",seconds_per_day // seconds_per_hour)
print ("Each answer agrees with the other.")
```
