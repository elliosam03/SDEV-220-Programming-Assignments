```python
#Sort an Array Exercise
class Solution:
    def sort012(self, arr):
        lo = 0
        mid = 0
        hi = len(arr) - 1

        while mid <= hi:
            if arr[mid] == 0:
                arr[lo], arr[mid] = arr[mid], arr[lo]
                lo = lo + 1
                mid = mid + 1

            elif arr[mid] == 1:
                mid = mid + 1

            else:
                arr[mid], arr[hi] = arr[hi], arr[mid]
                hi = hi - 1

        return arr
```

```python
#Binary Search Exercise
class Solution:
    def binarysearch(self, arr, k):
        l, r = 0, len(arr) - 1
        result = -1

        while l <= r:
            mid = l + (r - l) // 2

            if arr[mid] == k:
                result = mid
                r = mid - 1
            elif arr[mid] < k:
                l = mid + 1
            else:
                r = mid - 1

        return result
```
